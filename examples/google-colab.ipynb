# Using ngrok with Google Colab

Expose a Flask app running in a Colab notebook with the [ngrok-python](https://github.com/ngrok/ngrok-python) SDK.

1. Create an authtoken at https://dashboard.ngrok.com/get-started/your-authtoken
2. In Colab, open **Secrets** (key icon) and add `NGROK_AUTHTOKEN`
3. Run the cells below

```python
!pip install ngrok flask
```

```python
import os
import ngrok
from flask import Flask

# Colab: add NGROK_AUTHTOKEN via the Secrets sidebar (key icon).
# https://dashboard.ngrok.com/get-started/your-authtoken
try:
    from google.colab import userdata

    os.environ["NGROK_AUTHTOKEN"] = userdata.get("NGROK_AUTHTOKEN")
except ImportError:
    pass  # local Jupyter: export NGROK_AUTHTOKEN in your shell

if not os.environ.get("NGROK_AUTHTOKEN"):
    raise RuntimeError(
        "Set NGROK_AUTHTOKEN (Colab Secrets or your environment) before running."
    )

app = Flask(__name__)


@app.route("/")
def hello():
    return "Hello from Google Colab!"


listener = ngrok.forward(5000, authtoken_from_env=True)
print(f"\n * Public URL: {listener.url()}\n")

app.run(port=5000)
```
